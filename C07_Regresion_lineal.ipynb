# Regresiones lineales

La **regresión lineal** busca una relación lineal entre una variable continua (la que queremos predecir) y varias explicativas. Si los datos fueran una nube de puntos, la regresión intenta trazar la mejor recta posible que resuma la tendencia. 

Matemáticamente, esto significa que **asumimos** (pues no tiene por qué ser así) que si $X_1,X_2,...,X_p$ son nuestras columnas, aunque se trate de variables codificadas mediante técnicas como OneHot Encoding u Ordinal Encoding de modo que todas las columnas puedan operar numéricamente dentro del modelo, y $y$ es la columna objetivo, entonces existen números $\beta_0,\beta_1,...,\beta_p$ tales que $$y=\beta_0+\textcolor{red}{\underbrace{\beta_1X_1+\beta_2X_2+...+\beta_pX_p}_{\mbox{combinación lineal}}}+\varepsilon$$

donde permitimos un factor aleatorio $\varepsilon$ que no podemos controlar, pero del cual sabemos que $E[\varepsilon]=0$ y $Var(\varepsilon)=\sigma^2$.

A la ecuación anterior se le conoce como **modelo poblacional**.

Ahora bien, como **nunca podremos conocer los parámetros poblacionales** (y justo por eso hacemos inferencia estadística), tenemos que conformarnos con *estimaciones* de dichos parámetros. Si $\hat{\beta}_0,\hat{\beta}_1,...\hat{\beta}_p$ son las estimaciones que hacemos, entonces formamos el **modelo estimado**: $$\hat{y}=\hat{\beta}_0+\hat{\beta}_1X_1+\hat{\beta}_2X_2+...+\hat{\beta}_pX_p$$

Entender la diferencia entre ambos modelos es fundamental: $\hat{y}$ es el valor estimado una vez que elegimos $\hat{\beta}_0,\hat{\beta}_1,...\hat{\beta}_p$ en tanto que $y$ son los valores verdaderos de nuestros datos. De esta manera, el error que cometemos al proponer las $\hat{\beta_i}$ es $e=y-\hat{y}$ y por lo tanto $$y=\hat{\beta_0}+\hat{\beta_1}X_1+\hat{\beta_2}X_2+...+\hat{\beta_p}X_p+e$$

por lo que $e$, los errores cometidos, es la versión muestral de $\varepsilon$ (el factor aleatorio verdadero que nunca podremos controlar).

Una vez entendido lo anterior, recordemos que tenemos dos pasos:

**Paso 1.** **Suponemos que el fenómeno poblacional obedece a una regresión lineal**. Es decir, el fenómeno en toda la población se comporta como el **modelo poblacional**. Por lo tanto **debemos fundamentar esa suposición**.

**Paso 2.** **Nosotros proponemos las estimaciones** $\hat{\beta}_0,\hat{\beta}_1,...\hat{\beta}_p$. Esos estimadores dependen totalmente del investigador. Y por lo tanto los errores cometidos dependen de dichas propuestas. Y entonces buscamos que los errores cometidos sean lo más pequeño posible (recuerda que $y$ es una columa numérica).

Antes de pasar a resolver el **Paso 1**, resolveremos el **Paso 2**. Como dijimos, nosotros propondremos los estimadores y, una vez propuestos, podemos calcular los errores. Como buscamos que los errores sean lo mas pequeño posible, nuestra propuesta de estimadores debe cumplir este objetivo. Resolveremos este problema con una técnica conocida como *mínimos cuadrados*. 

Esta consiste en encontrar los estimadores $\hat{\beta}_0,\hat{\beta}_1,...\hat{\beta}_p$ de tal manera que $$\sqrt{\sum_{i=1}^n e_i^2}\mbox{ es mínima}$$ donde $e_i$ es el error cometido en el renglón $i$.

Si recordamos que $e_i=y_i-\hat{y}_i$, entonces buscamos minimizar $$\sqrt{\sum_{i=1}^n (y_i-\hat{y}_i)^2}$$ variando $\hat{\beta}_0,\hat{\beta}_1,...\hat{\beta}_p$

De esta manera, en términos de Álgebra Lineal, si $\hat{\beta}=(\hat{\beta}_0,\hat{\beta}_1,...,\hat{\beta}_p)$, queremos resolver el problema $$\min_{\hat{\beta}}\|y-\hat{y}(\hat{\beta})\|^2_2$$

Por otra parte, digamos que $X$ es la tabla formada por las columnas $X_1,X_2,...,X_p$. Es decir, $X=[X_1|X_2|...|X_p]$. Recuerda que esos son datos fijos, y el modelo muestral toma la forma $$\hat{y}(\hat{\beta})=\hat{\beta}_0+X\beta^\prime$$ donde $$\beta^\prime=\begin{bmatrix}\hat{\beta}_1\\\hat{\beta}_2\\\vdots\\\hat{\beta}_p\end{bmatrix}$$

Si añadimos una columna de 1's al inicio de $X$, una ventaja es que el problema se escribe únicamente en términos de $\hat{\beta}$. Esto es $$\hat{y}(\hat{\beta})=[1|X]\hat{\beta}$$ y por lo tanto buscamos resolver $$\min_{\hat{\beta}}\|y-[1|X]\hat{\beta}\|^2_2$$

## Las matemáticas detrás de la regresión lineal

Veamos las matemáticas que nos resuelven el problema de optimización anterior. 

**Teorema 1.** Sea $H$ un espacio con producto interno $\emptyset\neq M\subseteq H$ un subconjunto convexo y completo. Entonces para cada $u\in H$ existe un único $\tilde{u}\in M$ tal que $$\|u-\tilde{u}\|^2=\inf_{v\in M}\|u-v\|^2$$

**En cristiano.** En un universo donde se puedan medir ángulos y distancias, si $M$ es un conjunto suficientemente bonito y $u$ es un elemento de dicho universo, entonces existe un elemento $\tilde{u}$ en el conjunto que es el más cercano a $u$

**Teorema 2.** Bajo la misma notación que el teorema anterior, si $M$ es subespacio vectorial, entonces $u-\tilde{u}\in M^\perp$

**En cristiano.** si además de muy bonito, el conjunto $M$ es plano, entonces la recta que une a $u$ con $\tilde{u}$ es perpendicular a todo elemento de $M$

**Teorema 3.** Si $A$ es una matriz que transforma al espacio vectorial $H$ en $K$, entonces $$Im(A)^{\perp}=Ker(A^T)$$

**Corolario: Ecuaciones normales** Los estimadores $\hat{\beta}$ que resuelven $$\min_{\hat{\beta}}\|y-[1|X]\hat{\beta}\|^2_2$$ cumplen $$[1|X]^T[1|X]\hat{\beta}=[1|X]^Ty$$

*Demostración.* Queremos hallar $\hat{\beta}\in\mathbb{R}^{p+1}$ que minimice a $\|y-[1|X]\hat{\beta}\|^2_2$. Recordemos que $[1|X]$ ya es una matriz fija con $n$ filas y $p+1$ columnas. Además $y\in\mathbb{R}^n$. Por lo tanto, para cada $\hat{\beta}$ tendremos $[1|X]\hat{\beta}\in \mathbb{R}^n$. Esto es: $[1|X]$, visto como matriz, transforma $\mathbb{R}^{p+1}$ en $\mathbb{R}^n$.
 
Sea $M=Im([1|X])\subseteq\mathbb{R}^n$. Como $y\in\mathbb{R}^n$, por los Teoremas 1 y 2, existe $\tilde{y}\in M$ tal que $$\|y-\tilde{y}\|^2=\inf_{v\in M}\|y-v\|^2=\inf_{\hat{\beta}}\|y-[1|X]\hat{\beta}\|^2.$$ Esto resuelve el problema de optimización, y además, para este $\hat{\beta}$ se cumple $y-[1|X]\hat{\beta}\in M^\perp$. Ahora, por el Teorema 3, $$y-[1|X]\hat{\beta}\in Im([1|X])^\perp=Ker([1|X]^T)$$

Por lo tanto $$0=[1|X]^T(y-[1|X]\hat{\beta})=[1|X]^Ty-[1|X]^T[1|X]\hat{\beta}$$ de donde $$[1|X]^T[1|X]\hat{\beta}=[1|X]^Ty$$

**Teorema 4.** Sea $A$ una matriz donde todas las columnas son linealmente independientes. Entonces $A^TA$ es invertible.

*Demostración.* Sea $v$ tal que $A^TAv=0$. Multiplicando por $v^T$ a izquierda tenemos $$0=v^TA^TAv=(Av)^T(Av)=\|Av\|^2$$

Por lo tanto $Av=0$ y como todas la columnas de $A$ son linealmente independientes, la única manera en que $Av=0$ tenga solución es cuando $v=0$. De esta forma, $A^TA$ es inyectiva y, al ser cuadrada, es biyectiva. 

**Corolario 5.** Si $A$ tiene una columna que es combinación lineal de otras de sus columnas, entonces el sistema $$A^TAv=b$$ tiene infinitas soluciones.

**Regresión lineal en Python**

Veamos ahora cómo se relacionan las matemáticas anteriores con nuestro trabajo. Obsrvemos el **Corolario Importante** y el **Teorema 4**: si en $[1|X]$ las columnas son linealmente indpendientes (es decir, ya no hay problema de multicolinealidad), entonces los estimadores $\hat{\beta}=(\hat{\beta_0},\hat{\beta}_1,...,\hat{\beta}_p)$ que debemos proponer para el modelo muestral son $$\hat{\beta}=([1|X]^T[1|X])^{-1}[1|X]^Ty$$

Pero recordemos nuestro esquema de trabajo previo: tomamos nuestros datos originales, partimos en entrenamiento y prueba, separamos numéricas de categóricas, codificamos las categóricas, escalamos las numéricas, estandarizamos las numéricas y aplicamos PCA, de lo cual obtenemos los csv ``T_train_final_objetivo.csv`` y ``T_test_final_objetivo.csv`` 

El problema es el siguiente: si en nuestros *datos crudos* aparece una alguna columna nominal de $k$ categorías, al codificar tendremos $k$ columnas OneHot. Por lo tanto, en $[1|X]$ estarán estas columnas OneHot, y cumplen que **la suma de dichas columnas siempre es 1**. Como consecuencia, en $[1|X]$ tendremos una columna (la de 1's) que es combinación de lineal de las $k$ columnas OneHot que procesamos y entonces las ecuaciones normales no tienen solución única.  

En resumen: si hay alguna columna de tipo nominal, al aplicar el algoritmo ``PCA_full.ipynb`` obtendremos la tabla ``T_train_final_objetivo.csv`` a la cual **no le podremos aplicar regresión lineal**. En vista de esto, deberemos realizar un paso extra.


## Evaluación de modelos de regresión

Cuando el objetivo es **predecir**, las métricas responden dos preguntas distintas:  
1) **¿Qué tanto explico?** (panorámica)  
2) **¿Cuánto me equivoco?** (operativo, en unidades de la variable)  
Conviene reportar ambas.

### Error Absoluto Medio (MAE)

El **Error Absoluto Medio** cuantifica el **desvío típico** entre realidad y predicción en las mismas unidades de $y$. 
$$
\mathrm{MAE}=\frac{1}{n}\sum_{i=1}^{n}\lvert y_i-\hat y_i\rvert.
$$
Se interpreta de forma directa (pesos, días, grados, etc.). Es **menos sensible** a valores atípicos que el RMSE y, por eso, suele dar una idea estable del error “normal”. Si se requiere máxima robustez, la **mediana** del error absoluto (MedAE) es aún más inmune a outliers.

### Error Cuadrático Medio (MSE) y su raíz (RMSE)

El **MSE** promedia **errores al cuadrado**:
$$
\mathrm{MSE}=\frac{1}{n}\sum_{i=1}^{n}(y_i-\hat y_i)^2,
$$
por lo que queda en **unidades al cuadrado** (difícil de interpretar). Por eso se usa casi siempre su **raíz**:

$$
\mathrm{RMSE}=\sqrt{\frac{1}{n}\sum_{i=1}^{n}(y_i-\hat y_i)^2}.
$$

El **RMSE** vuelve a unidades de $y$ pero **penaliza más** los errores grandes; es el termómetro adecuado cuando los “patinones” importan especialmente (por ejemplo, pérdidas financieras fuertes por pocos casos mal predichos).

### Coeficiente de determinación ($R^2$) y $R^2$ ajustado

El **$R^2$** mide la **proporción de varianza** de $y$ que el modelo explica respecto a predecir siempre la media:
$$
R^2 = 1-\frac{\sum_{i}(y_i-\hat y_i)^2}{\sum_{i}(y_i-\bar y)^2}.
$$
Con **intercepto** en el modelo (que es como nosotros lo hacemos), $R^2$ coincide con $\mathrm{corr}(y,\hat y)^2$. Puede ser **negativo** (peor que la media) y **no** habla de magnitud del error ni garantiza correcta especificación.

El **$R^2$ ajustado** penaliza la complejidad:
$$
R^2_{\text{aj}} = 1-\frac{(1-R^2)(n-1)}{n-k-1},
$$
donde $n$ es el número de observaciones y $k$ el número de predictores (sin intercepto). Es útil para comparar **modelos con distinta cantidad de variables**, pero tampoco asegura buena forma funcional.

### NRMSE (RMSE Normalizado)

Para comparar entre problemas o datasets con distinta escala, conviene **normalizar**:
$$
\mathrm{NRMSE}=\frac{\mathrm{RMSE}}{\operatorname{sd}(y_{\text{test}})}.
$$
Así obtienes un error **relativo** (cuánto del “ancho natural” de $y$ queda como ruido). Como guía práctica, $\mathrm{NRMSE}\le 0.5$ suele indicar predicción **buena aquí dentro** (misma distribución que el set de prueba).

## Cómo decidir con un único test (sin Validación cruzada)

1. **Panorámica:** reporta $R^2_{\text{test}}$ para saber **qué tanto explicas**.  
2. **Operativo:** reporta **MAE** y **RMSE** (en unidades) para saber **cuánto te equivocas**.  
3. **Escala relativa:** añade **NRMSE** para comparar entre problemas y para un umbral claro.

**Regla de decisión (semáforo, “aquí dentro”):**  
- **VERDE (confiable):** $R^2_{\text{test}}\ge 0.70$ **y** $\mathrm{NRMSE}\le 0.50$.  
- **AMARILLO (usable con cautela):** $0.40\le R^2_{\text{test}}< 0.70$ o $0.50<\mathrm{NRMSE}\le 0.80$.  
- **ROJO (no confiable):** $R^2_{\text{test}}< 0.40$ o $\mathrm{NRMSE}> 0.80$.

> **Importante:** estas conclusiones valen **en el dominio del test**. No garantizan extrapolación fuera de rangos o poblaciones distintas.

## Plantilla de veredicto operativo

> **Veredicto.** El modelo **sí predice bien aquí dentro**: $R^2_{\text{test}}=\dots$, $\mathrm{MAE}=\dots$ (unidades de $y$), $\mathrm{RMSE}=\dots$ (unidades), $\mathrm{NRMSE}=\dots$.  

> **Criterio:** confiable si $R^2_{\text{test}}\ge 0.70$ **y** $\mathrm{NRMSE}\le 0.50$; aceptable con cautela en zona intermedia; no confiable si $R^2_{\text{test}}< 0.40$ **o** $\mathrm{NRMSE}> 0.80$.
