# Preprocesamiento

El preprocesamiento es la etapa previa al entrenamiento de un modelo en la que transformamos los datos crudos para que sean adecuados y comparables entre sí. Incluye tareas como rellenar valores faltantes, ajustar las escalas de las variables, codificar categorías y, en general, poner orden en la información.

Su objetivo es que los algoritmos puedan aprender de manera justa y eficiente, sin que una variable domine solo por tener números más grandes, sin que los huecos distorsionen los resultados y asegurando que tanto los datos de entrenamiento como los de prueba y producción se traten de forma consistente.

- Desde los 90s, al llevar ML al mundo real, se entendió que gran parte del trabajo es **limpiar y transformar** datos.  
- Los **pipelines** nacen para **evitar fugas**, **estandarizar procesos** y **reproducir** resultados.

### Objetivos de aprendizaje
- Separar correcta y tempranamente **entrenamiento** y **prueba**.  
- Tratar **valores faltantes** con estrategias **distintas por columna**.  
- **Escalar** variables numéricas con **escaladores diferentes por columna** (o *passthrough*).  
- **Codificar** variables categóricas (One-Hot).  
- Encadenar todo con **ColumnTransformer** y **Pipeline** para obtener un **dataset_final** reproducible.  
- **Persistir** el pipeline de preprocesamiento.
