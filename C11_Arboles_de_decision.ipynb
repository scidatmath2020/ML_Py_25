# Árboles de Decisión

Un árbol de decisión es un modelo que **aprende a clasificar objetos dentro de varias categorías posibles**, tomando decisiones paso a paso mediante preguntas sencillas.  

Por ejemplo: al analizar flores, un árbol puede preguntarse:  

> “¿El largo de pétalo mide menos de 2.5 cm?”  
> “¿La anchura del pétalo supera 1.8 cm?”  

Cada respuesta lleva a una nueva rama, y al final el árbol decide entre distintas clases: *setosa*, *versicolor*, *virginica*.  

En otras palabras: **un árbol no calcula fórmulas**, sino que **formula reglas lógicas** que van separando a las clases hasta obtener grupos homogéneos.

## Qué busca realmente el árbol

En cada punto donde debe dividir los datos, el árbol se pregunta:  
> “¿De qué forma puedo hacer dos grupos que sean lo más **puros** posible?”  

“Puro” significa: “donde la mayoría de los ejemplos pertenezca a una sola clase”.  
Si un grupo contiene flores de varias especies mezcladas en proporciones parecidas, el grupo está “impuro”.  
Si casi todas son de una sola especie, el grupo está “puro”.  

El árbol siempre busca la división que **reduzca al máximo la impureza**.

## Cómo mide la pureza cuando hay varias clases

Cuando existen varias categorías, el árbol calcula la pureza considerando **todas las proporciones simultáneamente**.  

Si en un grupo hay proporciones  
$$
(p_1, p_2, \dots, p_K)
$$
donde $p_k$ representa la fracción de la clase $k$, entonces define dos medidas comunes:  

- **Entropía:**  
  $$
  H = -\sum_{k=1}^{K} p_k \log_2 p_k
  $$
  mide el desorden global del nodo.  
  Si las clases están perfectamente mezcladas, la entropía es máxima.  

- **Índice de Gini:**  
  $$
  G = 1 - \sum_{k=1}^{K} p_k^2
  $$
  mide la probabilidad de clasificar mal una muestra si se elige al azar según las proporciones del nodo.  

Ambas son **cero cuando el nodo es puro** (una sola clase domina) y **máximas cuando las clases están mezcladas por igual**.

## El proceso de decisión

1. Se parte de todos los datos juntos en la raíz.  
2. En cada paso, el árbol evalúa todas las posibles preguntas del tipo:  
   “¿variable j < cierto umbral?”  
3. Calcula la reducción de impureza que cada pregunta produce.  
4. Elige la división que **más reduce la impureza total**.  
5. Repite el proceso dentro de cada nuevo grupo hasta que ya no haya ganancia significativa o se alcance una profundidad límite.  

Cada división mejora el orden: los nodos se van especializando en clases particulares.

## Interpretación geométrica

Geométricamente, cada pregunta **parte el espacio de datos con un plano perpendicular a una de las variables**.  

- En dos dimensiones, las regiones se vuelven rectángulos.  
- En tres, son prismas rectangulares.  
- En general, el espacio se va **fragmentando jerárquicamente**.  

Cada fragmento corresponde a una región del espacio donde **predomina una clase**.  
Así, el árbol va **pintando el plano de colores**: uno para cada categoría.

## La estructura del conocimiento

Cada ruta desde la raíz hasta una hoja se interpreta como **una regla de pertenencia**.  
Por ejemplo:

> Si la longitud del pétalo < 2.5 cm → clase = setosa  
> Si la longitud ≥ 2.5 cm y la anchura < 1.75 cm → clase = versicolor  
> Si la anchura ≥ 1.75 cm → clase = virginica  

Las reglas son **discretas y claras**, no continuas.  
Cada hoja contiene un grupo de observaciones mayoritariamente de una sola clase, y el modelo predice esa clase para nuevos ejemplos que cumplan las mismas condiciones.

## Profundidad y complejidad

Un árbol puede crecer hasta que cada hoja contenga apenas unos cuantos ejemplos.  
Eso genera **pureza perfecta en entrenamiento**, pero también **sobreajuste**: el modelo memoriza los datos.  

Para evitarlo, se imponen límites:  

- `max_depth` (profundidad máxima)  
- `min_samples_split` (mínimo de ejemplos para dividir un nodo)  
- `min_samples_leaf` (mínimo de ejemplos por hoja)  

Controlar estos parámetros es como **poda preventiva**: evita que el árbol se ramifique en exceso y pierda generalidad.

## Ventajas del enfoque multiclase

- Maneja cualquier número de categorías sin necesidad de entrenar varios modelos.  
- Genera reglas fácilmente interpretables.  
- Puede trabajar con variables numéricas o categóricas.  
- Detecta interacciones no lineales entre variables.

## Limitaciones

- Las fronteras de decisión son **rectangulares y discontinuas**.  
- Alta sensibilidad a los datos: un cambio pequeño puede alterar la estructura del árbol.  
- Con muchas clases pequeñas, puede generar ramas inestables.  

Por eso, para problemas complejos se suele usar un conjunto de árboles (bosques aleatorios o *boosting*), pero el **árbol individual** es la unidad conceptual y la herramienta de interpretación.

## El juez de categorías

Imagina un jurado de un concurso de ciencia.  
Cada juez evalúa proyectos y hace preguntas:  

> “¿Usa experimentos?”  
> “¿Tiene visualizaciones?”  
> “¿Incluye interacción?”  

Según las respuestas, el juez va encasillando los proyectos hasta decidir:  
> “Este es de la categoría *experimental*.”  
> “Este pertenece a *visualización interactiva*.”  
> “Este es de *comunicación oral*.”  

Ese juez está haciendo exactamente lo que hace un árbol de decisión multiclase: **formular preguntas jerárquicas hasta clasificar en una categoría única**.
