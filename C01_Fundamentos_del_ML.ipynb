## ¿Qué es el Machine Learning?
En términos prácticos, **ML es construir programas que mejoran su desempeño en una tarea a partir de datos**.  
Una forma clásica de expresarlo (parafraseando a Tom Mitchell, 1997): *dada una tarea **T**, un programa aprende de la experiencia **E** con una medida de desempeño **P**, si su desempeño en **T**, medido por **P**, mejora con **E***.

Por ejemplo: 
- Tarea *T*: identificar correos spam
- Desempeño *P*: exactitud
- Experiencia *E*: correos ya etiquetados como *spam* o *no spam*.

## Historia relámpago (contexto útil)
- 1950: Alan Turing plantea el **juego de la imitación** (test de Turing).
- 1959: Arthur Samuel usa ML para un programa de **damas** que aprende a jugar mejor con experiencia.
- 1960–80: primeras redes neuronales y el **perceptrón**; avanza la estadística computacional.
- 1990s: auge de **SVM** y **árboles**; mejora el desempeño con menos datos.
- 2006–2012: renacimiento del *deep learning*; 2012 **AlexNet** gana ImageNet con amplia ventaja.
- 2015–hoy: ML en productos cotidianos (traducción, visión, recomendaciones) y auge de pipelines reproducibles.

## Analogías para pensar ML
- **Niño estudiando para un examen.** El niño (el modelo) practica con guías (datos de entrenamiento), recibe retroalimentación (función de pérdida) y mejora sus respuestas (optimización). Si solo memoriza preguntas, le irá mal en el examen final (sobreajuste).
- **Entrenar a un perro.** Reforzamos conductas correctas con premios. En ML, los **parámetros** se ajustan para maximizar buen comportamiento frente a ejemplos.

## Tipos de aprendizaje
- **Supervisado:** tenemos ejemplos con etiqueta. Objetivos típicos: **clasificación** (clase discreta) y **regresión** (valor continuo).
- **No supervisado:** no hay etiquetas; buscamos **estructura** (agrupamientos, reducciones de dimensión).
- **Reforzamiento (mención):** un agente aprende a actuar por **recompensas** en un entorno. No lo veremos a detalle en este diplomado.

### Un conjunto de datos pequeño para jugar: Iris

El clásico dataset **Iris** (Fisher, 1936) contiene 150 flores de 3 especies con 4 mediciones. Lo convertiremos a un `DataFrame` para explorarlo.

```python
data = pd.read_csv("iris_dataset.csv")
data
```

### Visualización rápida
Una dispersión simple ayuda a intuir separaciones entre clases. No te preocupes por perfección estética todavía; la idea es mirar patrones.

```python
plt.figure()
for sp, subdf in data.groupby('Class'):
    plt.scatter(subdf['SepalLength'], subdf['PetalLength'], label=sp, alpha=0.8)
plt.xlabel('Sepal length (cm)')
plt.ylabel('Petal length (cm)')
plt.legend()
plt.title('Iris: vista rápida')
plt.show()
```

## Flujo típico de trabajo en ML
1. **Definir problema y métrica.** ¿Qué vamos a predecir y cómo mediremos el éxito?
2. **Preparar datos.** Limpieza, partición **entrenamiento/validación/prueba**, ingeniería de variables y **pipelines** reproducibles.
3. **Entrenar un modelo base.** Arranca simple; compara contra una línea base.
4. **Validar y ajustar.** Validación cruzada, selección de hiperparámetros y prevención de **fuga de datos**.
5. **Evaluar en prueba.** No se toca hasta el final.
6. **Despliegue y monitoreo.** Guardar modelos, reproducibilidad y mantenimiento.
