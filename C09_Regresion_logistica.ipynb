# Regresión logística.

Supongamos que tienes una moneda única y $n$ personas. Les pides que lancen la moneda. El resultado del lanzamiento dependerá de varios factores: estatura, sexo, edad, ángulo de lanzamiento, longitud de los dedos, etc. Es decir, depende de una serie de características de cada individuo.

Para cada persona anota dichas características en un renglón $\textbf{x}_i$ ($\textbf{x}_1$ son las características de la primera persona; $\textbf{x}_2$ son las características de la segunda persona; etc.) y por lo tanto obtienes una matriz $X$. 

Además, sea $y_i$ el lanzamiento obtenido en la persona $i$: 1 si es cara, 0 si es sello. 

Nuestro problema es "cuando llegue una nueva persona, anotarás sus características $x_{new}$ y quieres predecir lo que la moneda marcará". **En este capítulo resolveremos este problema utilizando el método conocido como regresión logística**.

## Las Matemáticas detrás 

### Función logística

Se define la **función logística estándar** como $$\sigma(t)=\frac{1}{1+e^{-t}}$$ donde $t$ es un número, y cuya gráfica es la siguiente
![imagenes](im015.png)

Notemos que $0<\sigma(t)<1$ para cualquier número $t$.

De esta manera, si $\textbf{t}=(t_1,t_2,t_3,...)$ definimos $\Sigma(\textbf{t})=\left(\sigma(t_1),\sigma(t_2),\sigma(t_3),...\right)$

### El modelo matemático

En la Regresión Logística, suponemos que en el modelo poblacional existen números $\beta_0, \beta_1,...,\beta_p$ tales que la probabilidad de la etiqueta 1 cumple que $$p=\Sigma(\beta_0+\beta_1X_1+\beta_2X_2+...+\beta_pX_p)=\Sigma([1|X]\beta)$$ 

Esto es: $$p_i=\sigma(\beta_0+\beta_1\mbox{caract 1 del individuo i}+\beta_2\mbox{caract 2 del individuo i}+...+\beta_p\mbox{caract p del individuo i})$$

En la ecuación anterior,  $$\beta_0+\beta_1\mbox{caract 1 del individuo i}+\beta_2\mbox{caract 2 del individuo i}+...+\beta_p\mbox{caract p del individuo i}$$ simplemente por $$\beta\bullet[1,\textbf{x}_i]$$ y por lo tanto $$p_i=\sigma(\beta\bullet[1,\textbf{x}_i])$$

### Distribución de $y_i$ dado el individuo $i$

Recordemos que $\textbf{x}_i$ son las características del individuo $i$ y $y_i$ es su etiqueta. Antes de tomar la medida $y_i$ (por ejemplo, lanzar el dado) el resultado es aleatorio, digamos $Y_i$. Entonces $Y_i|\textbf{x}_i$ (es decir, el resultado del lanzamiento dada la información del individuo) cumple que $Y_i|\textbf{x}_i\sim Bernoulli(p_i)$ donde $p_i$ es la probabilidad que tiene el individuo $i$ de obtener etiqueta 1. 

Por lo tanto $$P(Y_i=y_i|\textbf{x}_i)=p_i^{y_i}(1-p_i)^{1-y_i}$$

De esta manera, la función de verosimilitud de las probabilidades de todos los sujetos de la muestra es $$L(p_1,p_2,...,p_n)=\prod_{i=1}^np_i^{y_i}(1-p_i)^{1-y_i}$$



Pero también tomemos en cuenta que en regresión logística tenemos $p_i=\sigma(\beta\bullet[1,\textbf{x}_i])$

Por lo tanto, la función de verosimilitud en realidad depende de $\beta_0,\beta_1,...,\beta_p$ y por lo tanto, una vez que conocemos $X$ y $y$ (es decir, nuestra tabla completa), el investigador debe proponer $\hat{\beta}_0,\hat{\beta}_1,...,\hat{\beta}_p$ tales que $$L(\hat{\beta}_0,\hat{\beta}_1,...,\hat{\beta}_p)$$ sea máximo.

Finalmente, una vez hallado dicha colección $\hat{\beta}$ propuesta, cuando llegue un nuevo dato $\textbf{x}_{new}$ calcularemos su $p_i$. Si dicha probabilidad es mayor que un umbral $\alpha$ (también propuesta por el investigador) declararemos etiqueta 1 al renglón nuevo.
