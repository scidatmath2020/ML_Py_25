# DBSCAN (Density-Based Spatial Clustering of Applications with Noise)

K-means nos enseñó a pensar en clusters como regiones donde los puntos están “cerca” de un centroide.

Sin embargo, ese enfoque impone una forma geométrica rígida: cada grupo termina siendo esencialmente convexo y redondeado, y además todos los clusters tienen densidades similares. ¿Qué ocurre si los datos forman estructuras curvas o irregulares, como semicírculos, anillos o brazos entrelazados?
Ahí es donde DBSCAN (Density-Based Spatial Clustering of Applications with Noise) ofrece una alternativa más natural.

La idea es simple y poderosa: En lugar de preguntar “¿a qué centroide pertenece este punto?”, DBSCAN pregunta “¿hay suficiente densidad de puntos alrededor de este lugar?”.

Así, los clusters se definen como zonas densas separadas por regiones de baja densidad, y los puntos aislados se reconocen explícitamente como ruido.

## Idea geométrica: densidad y vecindarios

Cada punto tiene un vecindario de radio $\varepsilon$, y contamos cuántos puntos hay dentro. Si ese número supera un umbral (``min_samples``), decimos que el punto vive en una región suficientemente densa.

A partir de ese principio surgen tres tipos fundamentales de puntos:

- **Punto núcleo (core):** tiene al menos min_samples vecinos dentro de su radio $\varepsilon$.

- **Punto borde (border):** no tiene suficientes vecinos, pero está dentro del vecindario de un núcleo.

- **Ruido (noise):** ni es núcleo ni borde; vive en zonas aisladas o de baja densidad.

<img src="im030.png" width="50%" style="display: block; margin-left: auto; margin-right: auto;">

Los clusters se forman expandiendo regiones conectadas por puntos núcleo. Si dos núcleos están a distancia menor que $\varepsilon$, pertenecen al mismo cluster, y todos los bordes que se alcanzan desde allí también se agregan. El resultado son grupos con formas flexibles, que se adaptan al relieve real de los datos.

## Parámetros fundamentales

DBSCAN tiene solo dos parámetros principales:

- $\varepsilon$: el radio del vecindario.

    - Si $\varepsilon$ es muy pequeño, los puntos quedan aislados y surgen muchos grupos diminutos o ruido.

    - Si $\varepsilon$ es muy grande, regiones separadas se fusionan y todo se convierte en un único cluster.

<img src="im031.png" width="100%" style="display: block; margin-left: auto; margin-right: auto;">


- ``min_samples``: el número mínimo de puntos requeridos para que un punto sea núcleo.

    - Si es muy bajo, casi cualquier punto se convierte en núcleo.

    - Si es muy alto, incluso zonas densas se fragmentan.


<img src="im032.png" width="100%" style="display: block; margin-left: auto; margin-right: auto;">


Ambos parámetros controlan el nivel de granularidad: cuánto detalle queremos que DBSCAN considere como una región densa.

## Gráfico k-distancia: una herramienta práctica

Una forma empírica de elegir $\varepsilon$ consiste en calcular, para cada punto, la distancia a su k-ésimo vecino más cercano (donde k = ``min_samples`` − 1), y ordenar esas distancias.

Si graficamos el resultado, veremos una zona donde la curva crece de manera más brusca — ese “codo” indica el valor apropiado de ε: por debajo, los puntos viven en regiones densas; por encima, empieza el ruido.

## Comparación visual con K-means

DBSCAN es capaz de separar clusters con formas no convexas, incluso cuando se entrelazan. K-means, al basarse en distancias al centroide, corta el espacio con fronteras lineales y falla en estos casos.

Además, DBSCAN no necesita especificar el número de clusters; este se descubre automáticamente a partir de la densidad. También maneja el ruido de forma explícita, asignando etiqueta −1 a los puntos aislados.

## Limitaciones y consideraciones

El mayor desafío de DBSCAN es su sensibilidad cuando los datos presentan densidades muy distintas. Un único valor de $\varepsilon$ no puede adaptarse simultáneamente a regiones muy densas y muy dispersas: si $\varepsilon$ es bajo, los grupos dispersos se rompen; si es alto, los densos se fusionan.
