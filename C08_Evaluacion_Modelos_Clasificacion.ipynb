# Conceptos de Clasificación binaria

En clasificación binaria, tenemos el concepto de casos negativos (clase 0, en el caso del dataset de cancer de mama serian los casos donde el cancer es benigno) y casos positivos (clase 1, en el caso del dataset de cancer de mama serían los casos donde el cancer es maligno). Positivo y negativo no significa que el resultado sea bueno o malo, simplemente que la detección de un cancer maligno se active o no.

- Casos positivos: Casos donde la clase es 1 
- Casos negativos: Casos donde la clase es 0 

Esto nos lleva a computar 4 tipos de observaciones:

- **Verdaderos Positivos (True positives).** Etiquetas 1 que la computadora predice como 1.
- **Falsos Positivos (False positives).** Etiquetas 0 que la computadora predice como 1.
- **Verdaderos Negativos (True Negatives).** Etiquetas 0 que la computadora predice como 0.
- **Falsos Negativos (False Negatives).** Etiquetas 1 que la computadora predice como 0.

## Regla de decisión

En la mayoría de los modelos de clasificación (regresión logística, máquinas de soporte vectorial, árboles de decisión, K vecinos más cercanos), la regla para decidir a cuál de los cuatro tipos de observaciones pertenece cada predicción se basa en la selección de un **umbral de decisión**.

Esto es un número $\alpha$ que puede ser encontrado por la computadora o propuesto por el investigador. Básicamente, para renglón $i$ de nuestra tabla, el modelo de predicción elegido calculará un *score* $s_i$ y si $s_i\ge\alpha$, predecirá $\hat{y}_i=1$.

La selección del umbral $\alpha$ no es solo un parámetro técnico, sino una decisión estratégica que depende fundamentalmente de las consecuencias prácticas de los diferentes tipos de error. Por ejemplo, en el diagnóstico de enfermedades graves con tratamientos peligrosos:

- **Falso Positivo (FP):** Persona sana recibe tratamiento agresivo → Sufre efectos secundarios graves innecesariamente

- **Falso Negativo (FN):** Persona enferma no es detectada → La enfermedad progresa sin tratamiento

Para $\alpha$ alta (cercana a 1), es preferible no tratar a algunos enfermos (FN) que tratar mal a personas sanas (FP). En esta caso, los costos de los falsos positivos son inaceptablemente altos, por lo que solo diagnosticamos como positivos los casos donde tenemos certeza extrema (probabilidad > 0.99).

En este sentido, puedes pensar en $\alpha$ como "qué tan exigente soy para decir _positivo_".

Al mover $\alpha$ cambias cuántos casos predices como 1, y con eso se mueven los conteos.

**Regla general (al subir $\alpha$)**

- Predices menos positivos $\Rightarrow$
  - TP $\downarrow$ (no creciente)
  - FP $\downarrow$ (no creciente)
  - TN $\uparrow$ (no decreciente)
  - FN $\uparrow$ (no decreciente)

Al bajar $\alpha$ ocurre lo opuesto:

- Predices más positivos $\Rightarrow$ TP y FP suben; TN y FN bajan.

De hecho, en los casos extremos tenemos

- $\alpha > \max(score)$ $\rightarrow$ todo $0$: TP=0, FP=0, TN=#neg, FN=#pos.
- $\alpha \leq \min(score)$ $\rightarrow$ todo $1$: TP=#pos, FP=#neg, TN=0, FN=0.

### Ratios de clasificación

Una vez establecido $\alpha$, podemos calcular los TP, TN, FP, FN y calcular varias métricas de evaluación.

**Exactitud (Accuracy)**

La exactitud es una medida general de cómo se comporta el modelo; mide el porcentaje de casos que se han clasificado correctamente.

$$\mbox{Exactitud}=\frac{\mbox{Número de observaciones correctamente clasificadas}}{\mbox{Número de observaciones totales}}= \frac{VP+VN}{VP+VN+FP+FN}$$

**Precisión (Precision)**

De todas las predicciones positivas, ¿qué proporción son correctas?

$$\mbox{Precisión}=\frac{\mbox{Número de observaciones positivas correctamente clasificadas}}{\mbox{Número de observaciones clasificadas como positivas}}= \frac{VP}{VP+FP}$$

**Exhaustividad o sensibilidad(Recall o True Positive Rate)**

De todos los positivos reales, ¿qué proporción detectaste?

$$\mbox{Sensibilidad}=\frac{\mbox{Número de observaciones positivas clasificadas como positivas}}{\mbox{Número de observaciones positivas totales}}= \frac{VP}{VP+FN}$$

**Puntuación F1 (F1 score)**

La puntuación F1 es una media ponderada entre la sensibilidad (que intenta obtener cuantos mas verdaderos positivos independientemente de los falsos positivos) y la precisión (que intenta obtener solo verdaderos positivos que sean casos claros para limitar los falsos positivos).

La puntuación F1 se define como la media harmónica de la precisión y la sensibilidad:

$$F1=\frac{2}{\frac{1}{precisión}+\frac{1}{sensibilidad}}$$

## Regla práctica

- Si te importan ambas clases por igual y el dataset está razonablemente balanceado → mira Accuracy

- Si lo crítico es la clase positiva (fraude, cáncer) o hay desbalance fuerte → prioriza F1 para elegir umbral.

**Métricas que se heredan de $\alpha$**

- Recall/TPR = $\frac{TP}{(TP+FN)}$: $\downarrow$ al subir $\alpha$ (detectas menos positivos).
- Precisión = $\frac{TP}{(TP+FP)}$: suele $\uparrow$ al subir $\alpha$ (eres más estricto), pero no es estrictamente monótona: puede oscilar por empate/saltos.
- Accuracy: puede subir o bajar según el balance entre TN que ganas y TP que pierdes.
- F1: sube cuando precision y recall están equilibradas; no es monótona en $\alpha$.

**Intuición rápida**

- $\alpha$ bajo $\Rightarrow$ "generoso": casi todo es positivo $\rightarrow$ Recall alto, muchos FP.
- $\alpha$ alto $\Rightarrow$ "estricto": pocos positivos $\rightarrow$ Precision alta, Recall bajo.
